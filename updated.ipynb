```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
```

```python
from langchain_community.document_loaders import PyMuPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader=PyMuPDFLoader("Garvit _doc.pdf")
```

```python
docs=loader.load()
```

```python
print(len(docs))
```

```python
splitter=RecursiveCharacterTextSplitter(
     chunk_size=1000,
    chunk_overlap=150
)
```

```python
chunks=splitter.split_documents(docs)
```

```python
print(chunks[0])
```

```python
print(chunks[1])
```

```python
print(chunks[2])
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings=HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')
```

```python
from langchain_community.vectorstores import FAISS
```

```python
vector_store= FAISS.from_documents(
    chunks,
    embedding=embeddings
)

vector_store.save_local("vectorstorenew/faiss_index")
```

```python
print(type(vector_store))
```

```python
query = "What do you know about the candidate"
```

```python
result=vector_store.similarity_search(query, k=10)
```

```python
print(len(result))
```

```python
print(result[0].page_content)
```

```python
for i, doc in enumerate(result):
    print(f"\n========== RESULT {i+1} ==========")
    print(doc.page_content)
```

```python
vector_store.get_by_ids(['7f40f05f-4fa6-4a5a-8bee-bb905c4acd4c'])
```

```python
retriever = vector_store.as_retriever(
    search_type="mmr",
    search_kwargs={
        "k": 4, 
        "fetch_k": 10, # no of doc to consider befroe selection 
        "lambda_mult": 0.5 # how much relevance as compared to diversity
    }
)
```

```python
retrieved_docs=retriever.invoke(query)
```

```python
context_text= "\n\n".join(doc.page_content for doc in retrieved_docs)
```

```python
print(context_text)
```

```python
llm=ChatGroq(model="openai/gpt-oss-120b")
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
from langchain_core.runnables import RunnablePassthrough, RunnableLambda
```

```python
prompt = ChatPromptTemplate.from_template("""
You are KnowMe AI, a personal knowledge assistant for Garvit Pujari.

Answer the user's question using ONLY the context provided below.

Rules:
- Do not invent information.
- Do not add information that is not in the context.
- Do not create tables unless the user explicitly asks for a table.
- Answer naturally in concise paragraphs or bullet points.
- If multiple items are asked for, list them clearly.
- If the context contains the answer, use the available details.
- If the answer is not present in the context, say:
  "I don't have that information in my knowledge base."

Context:
{context}

Question:
{question}

Answer:
""")
```

```python
parser = StrOutputParser()
```

```python
print('j')
```

```python
def format_docs(docs):
    for doc in docs:
        return "\n\n".join(doc.page_content for doc in docs)
```

```python
rag_chain = (
    {
        "context": retriever | RunnableLambda(format_docs),
        "question": RunnablePassthrough()
    }
    | prompt
    | llm
    | parser
)
```

```python
rag_chain.invoke("What internships has Garvit completed?")
```
